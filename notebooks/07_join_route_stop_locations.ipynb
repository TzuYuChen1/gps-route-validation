# Import Required Libraries
Import pandas for data manipulation.

```python
# Import Required Libraries
import pandas as pd
```

# Load the CSV Files
Load route_stops_clean.csv and stop_locations_cleaned.csv into pandas DataFrames.

```python
# Load the CSV files
route_stops_df = pd.read_csv('<PROCESSED_DATA_DIR>/route_stops_clean.csv')
stop_locations_df = pd.read_csv('<PROCESSED_DATA_DIR>/stop_locations_cleaned.csv')
```

# Merge the DataFrames
Merge the two DataFrames on the 'stop_location_id' column using an inner join.

```python
# Merge the DataFrames on 'stop_location_id'
merged_df = pd.merge(route_stops_df, stop_locations_df, on='stop_location_id', how='left')
```

# Display the Merged Data
Display the first few rows of the merged DataFrame to verify the result.

```python
# Display the first few rows
merged_df.head()
```

# Save the Merged Data
Save the merged DataFrame to a new CSV file.

```python
# Save the merged DataFrame to a new CSV file
merged_df.to_csv('<PROCESSED_DATA_DIR>/route_stop_dim.csv', index=False)
print("Merged data saved to '<PROCESSED_DATA_DIR>/route_stop_dim.csv'")
```
