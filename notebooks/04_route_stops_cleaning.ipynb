```python
import pandas as pd
```

```python
path = "<RAW_DATA_DIR>/"
```

```python
route_stops_df = pd.read_csv(path + 'route_stops.csv')
route_stops_df.head()
```

```python
# Checking if our identifier is unique and not null
print(route_stops_df['route_stop_id'].is_unique)
print(route_stops_df['route_stop_id'].isna().sum())
```

```python
# Checking if all route_ids from route_stops are present in routes table
routes_df = pd.read_csv('<RAW_DATA_DIR>/routes.csv')
missing_route_ids = route_stops_df[~route_stops_df['route_id'].isin(routes_df['route_id'])]['route_id'].unique()
print("Missing route_ids:", missing_route_ids)
```

```python
# Checking if all stop_location_id from route_stops are present in stops table
stops_df = pd.read_csv('<RAW_DATA_DIR>/stop_locations.csv')
missing_stop_locations = route_stops_df[~route_stops_df['stop_location_id'].isin(stops_df['stop_location_id'])]['stop_location_id'].unique()
print("Missing stop_location_ids:", missing_stop_locations)
```

```python
# trip_type is already standardized
```

```python
# Classifying each stop
route_stops_df['is_school'] = route_stops_df['stop_types'].apply(lambda x: 'school' in x)
route_stops_df['is_anchor_stop'] = route_stops_df['stop_types'].apply(lambda x: 'anchor' in x)
route_stops_df['is_pickup'] = route_stops_df['stop_types'].apply(lambda x: 'pickup' in x)
route_stops_df['is_dropoff'] = route_stops_df['stop_types'].apply(lambda x: 'dropoff' in x)

# Assigning the final is_anchor value because there is anchor in is_anchor and stop_types
route_stops_df['final_anchor'] = route_stops_df['is_anchor'] | route_stops_df['is_anchor_stop']
```
