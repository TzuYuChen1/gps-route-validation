```python
import pandas as pd

# ── Paths Configuration ──────────────────────────────────────
PATH_TRIP_STOPS  = '<CLEANED_DATA_DIR>/trip_stops_long.csv'
PATH_ROUTE_DIM   = '<CLEANED_DATA_DIR>/route_stop_dim.csv'
PATH_OUTPUT      = '<CLEANED_DATA_DIR>/trip_stop_events.csv'

# ── Step 1: Read the two source tables ────────────────────────
trip_stops_long = pd.read_csv(PATH_TRIP_STOPS)
route_stop_dim  = pd.read_csv(PATH_ROUTE_DIM)

print("trip_stops_long shape :", trip_stops_long.shape)
print("route_stop_dim  shape :", route_stop_dim.shape)

# ── Step 2: Left join on route_stop_id ───────────────────────
# A left join (how='left') keeps all rows from trip_stops_long (105374 rows).
# If a route_stop_id has no corresponding record in route_stop_dim,
# the right table's columns will be filled with NaN.

trip_stop_events = trip_stops_long.merge(
    route_stop_dim,
    on='route_stop_id',
    how='left'
)

# ── Step 3: Save results ──────────────────────────────────────
trip_stop_events.to_csv(PATH_OUTPUT, index=False)
print(f"\nSaved to: {PATH_OUTPUT}")
```
